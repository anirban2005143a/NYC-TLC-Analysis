```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
clean_data = pd.read_parquet("../cleaned_data/clean_yellow_tripdata_2026-01.parquet")
df = clean_data.copy()
df.shape
```

# check payment_type

```python
df['payment_type'].value_counts(dropna=False)
```

# fill null payment type (if present) with most frequent type

```python
df['payment_type'] = df['payment_type'].fillna((df['payment_type'].mode()[0]))
```

# change data type

```python
df['payment_type'] = df['payment_type'].astype('int8')
```

```python
df.info()
```

# save back to the cleaned file 

```python
df.to_parquet('../cleaned_data/clean_yellow_tripdata_2026-01.parquet', index=False)
```
