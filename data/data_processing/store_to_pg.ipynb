```python
import pandas as pd
from sqlalchemy import create_engine, Interval
from sqlalchemy.engine import URL

connection_url = URL.create(
    drivername="postgresql+psycopg2",
    username="postgres",
    password="2005@BengaliSwim",
    host="localhost",
    port=5432,
    database="NYC_Taxi_Trip"
)

engine = create_engine(connection_url)

# Convert pandas Timedelta -> Python datetime.timedelta
df["trip_duration"] = pd.Series(
    [
        x.to_pytimedelta() if pd.notna(x) else None
        for x in df["trip_duration"]
    ],
    index=df.index,
    dtype="object"
)

print(df["trip_duration"].dtype)
print(type(df["trip_duration"].iloc[0]))
```

```python
df.to_sql(
    "yello_taxi_trip_2026_01",
    engine,
    if_exists="replace",
    index=False,
    dtype={"trip_duration": Interval()},
    chunksize=50_000,
    method="multi"
)

print("DataFrame stored successfully!")
```
